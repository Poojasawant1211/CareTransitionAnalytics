```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("/content/Wine Quality Dataset.csv")
```

```python
#"Fixed Acidity"
plt.figure(figsize=(8,6))
sns.histplot(data=data,x='fixed acidity',color="yellow",edgecolor="black",alpha=0.5,bins=5)
plt.title("Histogram of Fixed Acidity")
plt.xlabel("Fixed Acidity")
plt.ylabel("Count")
plt.show()
```

```python
#Calculate mean of fixed acidity
round(data['fixed acidity'].mean(),2)
```

```python
#Calculate the median of fixed acidity
data['fixed acidity'].median()
```

```python
plt.figure(figsize=(8,6))
sns.histplot(data=data,x='fixed acidity',color="yellow",edgecolor="black",alpha=0.5,bins=5)
plt.title("Histogram of Fixed Acidity")
plt.xlabel("Fixed Acidity")
plt.ylabel("Count")
plt.vlines(data['fixed acidity'].mean(),ymin=0,ymax=4000,color="green",label="Mean")
plt.vlines(data['fixed acidity'].median(),ymin=0,ymax=4000,color="red",label="Median")
plt.legend()
plt.show()
```

```python
#"Volatile Acidity"
plt.figure(figsize=(8,6))
sns.histplot(data=data,x='volatile acidity',color="skyblue",edgecolor="black",alpha=0.5,bins=5)
plt.title("Histogram of Volatile Acidity")
plt.xlabel("Volatile Acidity")
plt.ylabel("Density")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.distplot(data['volatile acidity'],color="green")
plt.title("Distplot of Volatile Acidity")
plt.xlabel("Volatile Acidity")
plt.ylabel("Density")
plt.show()
```

```python
#Calculate skewness
data['volatile acidity'].skew()
```

```python
#Calculate Mean
round(data['volatile acidity'].mean(),2)
```

```python
#Calculate Medain
data['volatile acidity'].median()
```

```python
plt.figure(figsize=(8,6))
sns.histplot(data=data,x='volatile acidity',color="skyblue",edgecolor="black",alpha=0.5,bins=5)
plt.title("Histogram of Volatile Acidity")
plt.xlabel("Volatile Acidity")
plt.ylabel("Density")
plt.vlines(data['volatile acidity'].mean(),ymin=0,ymax=3000,color="green",label="Mean")
plt.vlines(data['volatile acidity'].median(),ymin=0,ymax=3000,color="red",label="Median")
plt.legend()
plt.show()
```

```python
#Citric acid
plt.figure(figsize=(8,6))
sns.histplot(data=data,x='citric acid',color="pink",edgecolor="black",alpha=0.5,bins=5)
plt.title("Histogram of Citric Acidity")
plt.xlabel("Citric Acidity")
plt.ylabel("Count")
plt.show()
```

```python
#Quality Feature
plt.figure(figsize=(8,6))
sns.countplot(data=data,x='quality',color="orange")
plt.title("Countplot of Quality")
plt.xlabel("Quality")
plt.ylabel("Count")
plt.show()
```
